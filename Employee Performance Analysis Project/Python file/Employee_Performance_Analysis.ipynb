## Top 10 Employees performance Per Year

```python
print('Top 10 Employees performance Per Year')
Monthly_performance.groupby(['Employee_Id', 'Year'])\
    .agg(
        Total_Preformance=('Performance_Rating', 'sum'),
        Average_Performance=('Performance_Rating', 'mean'),
        Minimum_Performance = ('Performance_Rating', 'min'), 
        Maximum_Performance = ('Performance_Rating', 'max'),
        Months_Active=('Performance_Rating', 'count'))\
        .sort_values('Total_Preformance', ascending=False).head(10)
```

## Top 10 Employees AVG Satisfaction Per Year

```python
Monthly_performance.groupby(['Employee_Id', 'Year'])\
    .agg(
        Avg_satisfaction=('Employee_Satisfaction', 'mean'),
        Average_Performance=('Performance_Rating', 'mean'),
        Total_Absenteeism_days = ('Absenteeism_Days', 'sum'), 
        Avg_Benefits_Cost = ('Benefits_Cost', 'mean'))\
        .sort_values('Avg_satisfaction', ascending=False).head(10)
```

```python
Business_Outcomes = sheets['business_outcomes']
```

```python
Business_Outcomes.head()
```

```python
Business_Outcomes['Department'].value_counts()
```

```python
Business_Outcomes.groupby(['Department'])\
                    .agg(AVG_Target_Sales=('Sales_Target', 'mean'),
                         AVG_Actual_Sales=('Sales_Actual', 'mean'))\
                         .assign(
                             AVG_Gap = lambda x : x['AVG_Target_Sales'] - x['AVG_Actual_Sales'],
                             Target_status = lambda x : np.where(x['AVG_Gap'] >= 0, 'Achieved', 'Not Achieved')
                         )
```

```python
Business_Outcomes['Year'] = Business_Outcomes['Year_Month'].str[:4]
```

```python
Business_Outcomes.groupby(['Year','Department'])\
                    .agg(AVG_Target_Sales=('Sales_Target', 'mean'),
                         AVG_Actual_Sales=('Sales_Actual', 'mean'),
                         AVG_Waste_Percentage=('Waste_Percentage', 'mean'))\
                         .assign(
                             AVG_Gap = lambda x : x['AVG_Target_Sales'] - x['AVG_Actual_Sales'],
                             Target_status = lambda x : np.where(x['AVG_Gap'] >= 0, 'Achieved', 'Not Achieved')
                         )
```

```python
Business_Outcomes.groupby(['Year', 'Department'])\
                    .agg(AVG_Customer_Satisfaction=('Customer_Satisfaction', 'mean'),
                         AVG_NPS=('Nps_Score', 'mean'))
```

```python
Business_Outcomes.groupby(['Year_Month', 'Department'])\
                    .agg(AVG_Customer_Satisfaction=('Customer_Satisfaction', 'mean'),
                         AVG_NPS=('Nps_Score', 'mean'))\
                    .sort_values(['AVG_Customer_Satisfaction', 'AVG_NPS'], ascending=False).head(10)
```

```python
Business_Outcomes.groupby(['Year', 'Department'])\
                 .agg(
                     AVG_Cuatomer_Satisfaction=('Customer_Satisfaction', 'mean'),
                     AVG_On_Time_Delivery=('On_Time_Delivery', 'mean')
                     )
```
